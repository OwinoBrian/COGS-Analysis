#### `Date` Column

```python
copy_combined_time_logs['Date'] = pd.to_datetime(copy_combined_time_logs['Date'])
```

The above error is with regards to the attempted Datetime conversion we tried to effect in the above cell. The function `pd.to_datetime` is expecting a 'm/d/Y in the objects. The capital Y refers to the year in full, '2024'. Even though the `head()` function on the `combined_time_logs` shows that the dates in the `Date` column are in the format expected, '04/02/2024' some of the dates have the short year '4/2/24'.

Performing and in depth review of the data types in the column `Date`

```python
copy_combined_time_logs['Date'].apply(type).value_counts()
```

```python
copy_combined_time_logs['Date'].str.len().value_counts()
```

The source data has inconsistencies in how the dates have been entered. The first bucket i.e  dates with a length of 6 are in the format of m/d/y, the second bucker(7) are formated as m/dd/y while the last bucket (10) are in the format m/d/Y - Y capitalized means the year in full

```python
copy_combined_time_logs.loc[copy_combined_time_logs['Date'].str.len() == 6, 'Date'].head(10)

```

```python
copy_combined_time_logs.loc[copy_combined_time_logs['Date'].str.len() == 7, 'Date'].head(10)
```

```python

copy_combined_time_logs.loc[copy_combined_time_logs['Date'].str.len() == 10, 'Date'].head(10)
```

To avoid errors associated with inferencing with `copy_combined_time_logs['Date'] = pd.to_datetime(copy_combined_time_logs['Date'])` as the code output, the next steps will involve updating the `Date` column in their various bucket to allow for a more a controlled approached. 

```python
# This is to remove any leading or trailing spaces in the 'Date' column
copy_combined_time_logs['Date'] = (copy_combined_time_logs['Date']
    .astype(str)
    .str.strip()
)
```

```python
# Creating masks for different lengths of the 'Date' strings
mask_6  = copy_combined_time_logs['Date'].str.len() == 6
mask_7  = copy_combined_time_logs['Date'].str.len() == 7
mask_10 = copy_combined_time_logs['Date'].str.len() == 10
```

```python
copy_combined_time_logs.loc[mask_6, 'Date'] = pd.to_datetime(
    copy_combined_time_logs.loc[mask_6, 'Date'],
    format='%m/%d/%y',
    errors='coerce'
)

copy_combined_time_logs.loc[mask_7, 'Date'] = pd.to_datetime(
    copy_combined_time_logs.loc[mask_7, 'Date'],
    format='%m/%d/%y',
    errors='coerce'
)

copy_combined_time_logs.loc[mask_10, 'Date'] = pd.to_datetime(
    copy_combined_time_logs.loc[mask_10, 'Date'],
    format='%m/%d/%Y',
    errors='coerce'
)
```

```python
copy_combined_time_logs['Date'] = pd.to_datetime(copy_combined_time_logs['Date'])
```
