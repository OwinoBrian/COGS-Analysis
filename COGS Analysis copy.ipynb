#### 1.3.1 Categorical Columns

```python
categorical_columns = ['Status', 'Person > Department > Type', 'Person > Brand', 
                       'Location', 'Project', 'Bid Link', 'Task', 'Type', 
                       'Department', 'Step', 'Pay Period']

for col in categorical_columns:
    time_logs[col] = time_logs[col].astype('category')
```

```python
from datetime import datetime

def parse_date(date_str, formats):
    for fmt in formats:
        try:
            return datetime.strptime(date_str, fmt)
        except ValueError:
            pass
    return None

def sort_dates(date_strings, formats):
  return sorted(date_strings, key=lambda x: parse_date(x, formats))

date_strings = ["2024-01-01", "02/15/2023", "2023-12-25", "10-05-2022"]
date_formats = ["%Y-%m-%d", "%m/%d/%Y", "%Y-%m-%d", "%d-%m-%Y"]

sorted_dates = sort_dates(date_strings, date_formats)
print(sorted_dates)
# Expected Output: ['10-05-2022', '02/15/2023', '2023-12-25', '2024-01-01']
```

### 1.4 Checking and handling missing values

```python
missing_values = time_logs.isnull().sum()
print("Missing values in each column:")
print(missing_values)
```

Due to the nature of the columns that are missing values, we will fill the with "Unknwown"

```python
#Since the script had converted 'status', 'bid link',etc into categorical columns, We have to insert 'unknown' as a category
# List of categorical columns with missing values
categorical_columns_with_missing = ['Status', 'Bid Link', 'Task', 'Step', 'Pay Period']


for col in categorical_columns_with_missing:
    # Only add the "Unknown" category if it's not already present
    if 'Unknown' not in time_logs[col].cat.categories:
        time_logs[col] = time_logs[col].cat.add_categories('Unknown')
    
    # Fill missing values with "Unknown"
    time_logs[col].fillna('Unknown', inplace=True)

# Check for any remaining missing values
print(time_logs.isnull().sum())

```

## 2. Feature Engineering 

From the data set, it can be assummed that the entries in the `logged hours` column are the differences between their corresponding start times and stop times. The script will establish if this is the case

```python
# calculating the difference and adding the result to the new 'duration' column
time_logs['Duration'] = (time_logs['Stop Time'] - time_logs['Start Time']).dt.total_seconds() / 3600


print(time_logs[['Start Time', 'Stop Time', 'Duration', 'Logged Hours']].head())
```

###  2.1 Comparing 'Duration' to 'Logged Hours'

```python
# Check for discrepancies between Logged Hours and calculated Duration
discrepancies = time_logs[time_logs['Logged Hours'] != time_logs['Duration']]

# Get the total count of discrepancies
discrepancy_count = discrepancies.shape[0]

print(f"Total count of discrepancies: {discrepancy_count}")
```

Logged hours will be substituted with the new `duration` entires.

```python
time_logs['Logged Hours'] = time_logs['Duration']
```

#### 2.2.1 Weekly Aggregation

```python
# Add a Week column based on the Date
time_logs['Week'] = time_logs['Date'].dt.isocalendar().week

# Group by Week and Department, summing the Logged Hours
weekly_aggregation = time_logs.groupby(['Week', 'Department']).agg({
    'Logged Hours': 'sum'
}).reset_index()

# View a sample of the weekly aggregation
weekly_aggregation

```

```python
# Sort the weekly aggregation by Logged Hours in descending order
weekly_aggregation = weekly_aggregation.sort_values(by='Logged Hours', ascending=False)

# View the sorted aggregation (top departments by Logged Hours)
print(weekly_aggregation.head())
```

```python
# Count the number of employees (Person) in each Department per Week
weekly_employee_count = time_logs.groupby(['Week', 'Department'])['Id'].nunique().reset_index(name='Employee Count')

# Merge the employee count into the weekly aggregation
weekly_aggregation = pd.merge(weekly_aggregation, weekly_employee_count, on=['Week', 'Department'])

# Calculate Total Available Hours per department per week (40 hours per employee)
weekly_aggregation['Total Available Hours'] = weekly_aggregation['Employee Count'] * 40

# Calculate Utilization Rate: (Logged Hours / Total Available Hours) * 100
weekly_aggregation['Utilization Rate'] = (weekly_aggregation['Logged Hours'] / weekly_aggregation['Total Available Hours']) * 100

# Sort by Utilization Rate in descending order
weekly_aggregation = weekly_aggregation.sort_values(by='Utilization Rate', ascending=False)

# View the final weekly aggregation with Utilization Rate
print(weekly_aggregation[['Week', 'Department', 'Logged Hours', 'Employee Count', 'Total Available Hours', 'Utilization Rate']].head())
```

#### 2.2.2 Monthly Aggregation

```python
# Add a Month column based on the Date
time_logs['Month'] = time_logs['Date'].dt.to_period('M')

# Group by Month and Department, summing the Logged Hours
monthly_aggregation = time_logs.groupby(['Month', 'Department']).agg({
    'Logged Hours': 'sum'
}).reset_index()

# View a sample of the monthly aggregation


monthly_aggregation.sort_values(by='Logged Hours', ascending=False)
```

```python
# Count the number of unique employees (Person) in each Department per Month
monthly_employee_count = time_logs.groupby(['Month', 'Department'])['Id'].nunique().reset_index(name='Employee Count')

# Merge the employee count into the monthly aggregation
monthly_aggregation = pd.merge(monthly_aggregation, monthly_employee_count, on=['Month', 'Department'])

# View a sample of the merged data with employee count
monthly_aggregation.sort_values(by='Logged Hours', ascending=False)
```

```python
# Calculate Total Available Hours per department per month (160 hours per employee)
monthly_aggregation['Total Available Hours'] = monthly_aggregation['Employee Count'] * 160

# Calculate Utilization Rate: (Logged Hours / Total Available Hours) * 100
monthly_aggregation['Utilization Rate'] = (monthly_aggregation['Logged Hours'] / monthly_aggregation['Total Available Hours']) * 100

# View the final monthly aggregation with Utilization Rate
print(monthly_aggregation[['Month', 'Department', 'Logged Hours', 'Employee Count', 'Total Available Hours', 'Utilization Rate']].head())
```

```python
# Sort by Utilization Rate in descending order (highest utilization at the top)
monthly_aggregation_sorted = monthly_aggregation.sort_values(by='Utilization Rate', ascending=False)

# View a sample of the sorted data
print(monthly_aggregation_sorted.head())
```

## 3. Visualization

```python
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.express as px
import pandas as pd

# Sample data (replace 'weekly_aggregation' with your DataFrame)
# weekly_aggregation is your DataFrame after processing

# Initialize the Dash app
app = dash.Dash(__name__)

# Custom CSS styles for rounded outlines and box shadows
box_style = {
    'padding': '20px', 
    'margin': '10px',
    'box-shadow': '0 4px 8px rgba(0, 0, 0, 0.1)',
    'border-radius': '10px',
    'background-color': 'white',
    'border': '1px solid #ddd'
}

score_card_style = {
    'text-align': 'center',
    'padding': '20px',
    'margin': '10px',
    'box-shadow': '0 4px 8px rgba(0, 0, 0, 0.1)',
    'border-radius': '10px',
    'background-color': 'white',
    'border': '1px solid #ddd',
}

score_value_style = {
    'font-size': '24px',
    'font-weight': 'bold',
    'color': '#007BFF'  # Blue color for emphasis
}

# Dashboard Layout
app.layout = html.Div([
    html.Div([
        html.H1('COGS Utilization Analysis', style={'text-align': 'center', 'padding': '20px'}),
    ], style={**box_style, 'text-align': 'center'}),  # Title box shadow and rounded

    # Filters
    html.Div([
        html.Div([
            html.Label('Department'),
            dcc.Dropdown(
                id='department-dropdown',
                options=[{'label': 'Select All', 'value': 'ALL'}] + [{'label': dept, 'value': dept} for dept in weekly_aggregation['Department'].unique()],
                value='ALL',
                clearable=False,
                style={'width': '100%'}
            ),
        ], style={**box_style, 'width': '45%', 'display': 'inline-block'}),

        html.Div([
            html.Label('Week'),
            dcc.Dropdown(
                id='week-dropdown',
                options=[{'label': 'Select All', 'value': 'ALL'}] + [{'label': str(week), 'value': week} for week in weekly_aggregation['Week'].unique()],
                value='ALL',
                clearable=False,
                style={'width': '100%'}
            ),
        ], style={**box_style, 'width': '45%', 'display': 'inline-block'})
    ], style={'padding': '20px', 'display': 'flex', 'justify-content': 'space-around'}),

    # Scorecards
    html.Div([
        html.Div([
            html.H4("Average Work Hours/Week", style={'text-align': 'center'}),
            html.Div(id='avg-hours-week', style=score_value_style),
        ], style=score_card_style),

        html.Div([
            html.H4("Average Utilization Rate", style={'text-align': 'center'}),
            html.Div(id='avg-util-rate', style=score_value_style),
        ], style=score_card_style),

        html.Div([
            html.H4("Total Hours Worked", style={'text-align': 'center'}),
            html.Div(id='total-hours', style=score_value_style),
        ], style=score_card_style)
    ], style={'display': 'flex', 'justify-content': 'space-around', 'padding': '20px'}),

    # Second Section: Bar chart and Column chart side by side
    html.Div([
        html.Div([
            dcc.Graph(id='bar-graph')
        ], style={**box_style, 'width': '50%', 'display': 'inline-block'}),
        html.Div([
            dcc.Graph(id='column-graph')
        ], style={**box_style, 'width': '50%', 'display': 'inline-block'})
    ], style={'padding': '10px', 'display': 'flex', 'justify-content': 'space-around'}),

    # Third Section: Table and Pie chart side by side
    html.Div([
        html.Div([
            html.H4('Top 10 Departments by Utilization Rate', style={'text-align': 'center'}),
            html.Table(id='top-departments-table', style={'width': '100%', 'border': '1px solid #ccc'}),
        ], style={**box_style, 'width': '50%', 'display': 'inline-block', 'background-color': '#f9f9f9'}),
        html.Div([
            dcc.Graph(id='pie-chart')
        ], style={**box_style, 'width': '50%', 'display': 'inline-block'})
    ], style={'padding': '10px', 'display': 'flex', 'justify-content': 'space-around'})
], style={'background-color': '#f9f9f9', 'padding': '20px'})

# Callback to update scorecards, graphs, and table
@app.callback(
    [Output('avg-hours-week', 'children'),
     Output('avg-util-rate', 'children'),
     Output('total-hours', 'children'),
     Output('bar-graph', 'figure'),
     Output('column-graph', 'figure'),
     Output('top-departments-table', 'children'),
     Output('pie-chart', 'figure')],
    [Input('department-dropdown', 'value'),
     Input('week-dropdown', 'value')]
)
def update_dashboard(selected_department, selected_week):
    # Filter data by week (and department if not 'ALL')
    if selected_week == 'ALL':
        filtered_data_week = weekly_aggregation
    else:
        filtered_data_week = weekly_aggregation[weekly_aggregation['Week'] == selected_week]

    if selected_department == 'ALL':
        filtered_data_dept = filtered_data_week
    else:
        filtered_data_dept = filtered_data_week[filtered_data_week['Department'] == selected_department]

    # Check if filtered data is empty
    if filtered_data_dept.empty or filtered_data_week.empty:
        return (
            "No data available",
            "No data available",
            "No data available",
            {}, {}, [html.Tr([html.Td("No data available")])], {}
        )

    # Calculate key metrics for the selected department or all departments
    avg_hours_week = filtered_data_dept['Logged Hours'].mean()
    avg_util_rate = filtered_data_dept['Utilization Rate'].mean()
    total_hours = filtered_data_dept['Logged Hours'].sum()

    # Format numbers for scorecards
    avg_hours_week_text = f"{avg_hours_week:.2f}"
    avg_util_rate_text = f"{avg_util_rate:.2f}%"
    total_hours_text = f"{total_hours:.2f}"

    # Bar graph: Utilization rate per department (using data filtered by week)
    bar_fig = px.bar(
        filtered_data_week,
        x='Department',
        y='Utilization Rate',
        title='Utilization Rate by Department'
    )

    # Column chart: Total logged hours per department (using data filtered by week)
    column_fig = px.bar(
        filtered_data_week,
        x='Department',
        y='Logged Hours',
        title='Total Logged Hours by Department'
    )

    # Table: Top 10 departments by utilization rate (using data filtered by week)
    top_departments = filtered_data_week.nlargest(10, 'Utilization Rate')[['Department', 'Utilization Rate']]
    
    # Table rows with light blue banded style
    table_rows = [
        html.Tr([html.Th('Department'), html.Th('Utilization Rate')], style={'background-color': '#e6f7ff'})
    ]
    for i, (_, row) in enumerate(top_departments.iterrows()):
        row_color = '#f2f9ff' if i % 2 == 0 else '#ffffff'
        table_rows.append(html.Tr([
            html.Td(row['Department']),
            html.Td(f"{row['Utilization Rate']:.2f}%")
        ], style={'background-color': row_color}))

    # Pie chart: Logged hours vs. unutilized hours for the selected department
    total_logged_hours = filtered_data_dept['Logged Hours'].sum()
    total_available_hours = filtered_data_dept['Employee Count'].sum() * 40  # Assuming 40 hours/week
    unutilized_hours = total_available_hours - total_logged_hours
    pie_fig = px.pie(
        values=[total_logged_hours, unutilized_hours],
        names=['Logged Hours', 'Unutilized Hours'],
        title='Logged vs Unutilized Hours'
    )

    return avg_hours_week_text, avg_util_rate_text, total_hours_text, bar_fig, column_fig, table_rows, pie_fig

# Run the Dash app and open in the browser
if __name__ == '__main__':
    app.run_server(debug=True, use_reloader=False)
```
