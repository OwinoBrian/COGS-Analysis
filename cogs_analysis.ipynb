```r
# Load individual time logs for April, May, and June
april_logs <- read.csv("C:\\Users\\mukam\\OneDrive\\Documents\\Projects\\COGS Analysis\\timelogs\\timelogs\\timelogs_april.csv")
may_logs <- read.csv("C:\\Users\\mukam\\OneDrive\\Documents\\Projects\\COGS Analysis\\timelogs\\timelogs\\timelogs_may.csv")
june_logs <- read.csv("C:\\Users\\mukam\\OneDrive\\Documents\\Projects\\COGS Analysis\\timelogs\\timelogs\\timelogs_june.csv")

# Combine all time logs into one data frame
time_logs <- rbind(april_logs, may_logs, june_logs)

# View the combined data
head(time_logs)
```

```r
# Convert Date and Time columns to appropriate formats
time_logs$Date <- as.Date(time_logs$Date, format = "%m/%d/%Y")
time_logs$Start_Time <- as.POSIXct(time_logs$Start.Time, format = "%m/%d/%Y %H:%M")
time_logs$Stop_Time <- as.POSIXct(time_logs$Stop.Time, format = "%m/%d/%Y %H:%M")
```

```r
# Data Visualisation - At this point we are going to use R Shiny to set up the environment and dashboard layout
```

```r
##`shiny`,`ggplot` and `dplyr` are the packages the script will be using to build the application, create plots and manipulate the data for analysis
```

```r
library(shiny)
library(ggplot2)
library(dplyr)
```

```r
# UI-Interface - The dashboard will be split into four sections
### Title,
### Sidebar Layout
### Data Range Input
### Output
```

```r
# UI - User Interface
ui <- fluidPage(
  titlePanel("COGS Resource Utilization Dashboard"),
  
  sidebarLayout(
    sidebarPanel(
      # Input to select time range (optional)
      dateRangeInput("dateRange", "Select Date Range:",
                     start = min(time_logs$Date), end = max(time_logs$Date))
    ),
    
    mainPanel(
      # Output the visuals
      plotOutput("utilizationPlot"),      # Plot for weekly utilization
      plotOutput("barPlot")               # Another visual (e.g., monthly aggregation)
    )
  )
)
```

```r
# Server - Logic and visuals generation
server <- function(input, output) {
  
  # Reactive expression to filter data based on date range input
  filtered_data <- reactive({
    time_logs %>%
      filter(Date >= input$dateRange[1] & Date <= input$dateRange[2])
  })
  
  # Line chart: Weekly utilization trend
  output$utilizationPlot <- renderPlot({
    weekly_data <- filtered_data() %>%
      group_by(week = as.numeric(format(Date, "%U"))) %>%
      summarise(utilization_hours = sum(as.numeric(difftime(Stop_Time, Start_Time, units = "hours"))))
    
    ggplot(weekly_data, aes(x = week, y = utilization_hours)) +
      geom_line(color = "blue", linewidth = 1.2) +
      geom_hline(yintercept = 40 * 5 * 0.8, linetype = "dashed", color = "red") + # Target line (80% of 40 hours per week)
      labs(title = "Weekly COGS Utilization", x = "Week", y = "Total Hours Utilized") +
      theme_minimal()
  })
  
  # Bar plot: Monthly utilization
  output$barPlot <- renderPlot({
    monthly_data <- filtered_data() %>%
      mutate(month = format(Date, "%B")) %>%
      group_by(month) %>%
      summarise(utilization_hours = sum(as.numeric(difftime(Stop_Time, Start_Time, units = "hours"))))
    
    ggplot(monthly_data, aes(x = month, y = utilization_hours)) +
      geom_bar(stat = "identity", fill = "skyblue") +
      labs(title = "Monthly COGS Utilization", x = "Month", y = "Total Hours Utilized") +
      theme_minimal()
  })
}

# Run the application
shinyApp(ui = ui, server = server)
```
